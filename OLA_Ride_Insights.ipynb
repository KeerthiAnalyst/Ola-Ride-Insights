```python
# Create a copy of the original dataset
df_clean = df.copy()

# Standardize text values safely
for col in df_clean.columns:
    df_clean[col] = df_clean[col].map(
        lambda x: x.strip() if isinstance(x, str) else x
    )

# Convert Date column into datetime format
df_clean['Date'] = pd.to_datetime(
    df_clean['Date'],
    errors='coerce'
)

# Check results
print("Original Shape:", df.shape)
print("Cleaned Shape:", df_clean.shape)

print("Duplicate Rows:", df_clean.duplicated().sum())

print("\nData Types:")
print(df_clean.dtypes)

print("\nMissing Values:")
print(df_clean.isnull().sum().sum())
```

```python
df_clean.to_csv("OLA_Cleaned_Data.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python
df = pd.read_csv("OLA_Cleaned_Data.csv")

sns.set_style("whitegrid")

print("Dataset Shape:", df.shape)
```

```python
# Convert Date into proper datetime format
df['Date'] = pd.to_datetime(df['Date'])

# Extract booking date without time
df['Booking_Date'] = df['Date'].dt.date

# Count bookings per day
daily_bookings = df.groupby('Booking_Date')['Booking_ID'].count()

# Plot
plt.figure(figsize=(14, 5))
daily_bookings.plot(kind='line', marker='o')

plt.title('Daily Ride Booking Trends - July 2024')
plt.xlabel('Date')
plt.ylabel('Number of Bookings')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
booking_status = df['Booking_Status'].value_counts()

plt.figure(figsize=(8, 5))
booking_status.plot(kind='bar', color='skyblue')

plt.title('Booking Status Breakdown')
plt.xlabel('Booking Status')
plt.ylabel('Number of Bookings')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
print(
    (df['Booking_Status'].value_counts(normalize=True) * 100).round(2)
)
```

```python
vehicle_bookings = df['Vehicle_Type'].value_counts()

plt.figure(figsize=(10, 5))
vehicle_bookings.plot(kind='bar', color='orange')

plt.title('Bookings by Vehicle Type')
plt.xlabel('Vehicle Type')
plt.ylabel('Total Bookings')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
successful_rides = df[df['Booking_Status'] == 'Success']

revenue_payment = successful_rides.groupby(
    'Payment_Method'
)['Booking_Value'].sum().sort_values(ascending=False)

plt.figure(figsize=(9, 5))
revenue_payment.plot(kind='bar', color='green')

plt.title('Revenue by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Total Booking Value')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
top_locations = df['Pickup_Location'].value_counts().head(10)

plt.figure(figsize=(12, 5))
top_locations.plot(kind='bar', color='purple')

plt.title('Top 10 Pickup Locations')
plt.xlabel('Pickup Location')
plt.ylabel('Number of Bookings')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=successful_rides,
    x='Driver_Ratings',
    bins=10,
    kde=True
)

plt.title('Driver Ratings Distribution')
plt.xlabel('Driver Rating')
plt.ylabel('Frequency')
plt.show()
```

```python
rating_comparison = successful_rides[
    ['Driver_Ratings', 'Customer_Rating']
].mean()

print(rating_comparison)

rating_comparison.plot(
    kind='bar',
    figsize=(7, 5),
    color=['green', 'orange']
)

plt.title('Average Driver vs Customer Ratings')
plt.ylabel('Average Rating')
plt.show()
```
